#### bs4 library Beautifulsoup module

```python
import sys, subprocess

try:
    from bs4 import BeautifulSoup as bs
except ModuleNotFoundError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "beautifulsoup4"])
    from bs4 import BeautifulSoup as bs
```

```python
clean_data=bs(data1,"xml")
```

```python
print(clean_data)
```

```python
#to find multiple tags we use find_all() 
title_tgs=clean_data.find_all("title")
title_tgs
```

```python
for i in title_tgs:
    print(i.get_text())
```

```python
#to find single or first tag we use find() 
data2=clean_data.find("title")
data2
```

```python
data2.get_text()
```

```python
year=clean_data.find("year")
```

```python
year
```

```python
year.get_text()
```

```python
year=clean_data.find_all("year")
year
```

```python
for y in year:
    print(y.get_text())
```
