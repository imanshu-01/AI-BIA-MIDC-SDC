# Calculator

```python
from tkinter import *
```

```python
window=Tk()
```

```python
def add():
    a=int(entry1.get())
    b=int(entry2.get())
    c=a+b
    result_plus.insert(0,c)
```

```python
def subtract():
    a=int(entry1.get())
    b=int(entry2.get())
    c=a-b
    result_sub.insert(0,c)
```

```python
def clear():
    entry1.delete(0,END)
    entry2.delete(0,END)
    result_plus.delete(0,END)
    result_sub.delete(0,END)
    
```

```python
label1=Label(window,text="Enter No.1: ", bg="blue", fg="white",padx=20,pady=20)
label2=Label(window,text="Enter No.2: ", bg="red", fg="white",padx=20,pady=20)

entry1=Entry(window)
entry2=Entry(window)
label1.grid(row=0,column=0)
label2.grid(row=1,column=0)
entry1.grid(row=0,column=1)
entry2.grid(row=1,column=1)
add=Button(window,text="add",command=add)
subtraction=Button(window,text="sub",command=subtract)
add.grid(row=2,column=0,padx=20,pady=20)
subtraction.grid(row=3,column=0,padx=20,pady=20)
result_plus=Entry(window)
result_plus.grid(row=2,column=1)
result_sub=Entry(window)
result_sub.grid(row=3,column=1)
clear=Button(window,text="clear all",command=clear)
clear.grid(row=4,column=0)
```

```python
window.mainloop()
```
