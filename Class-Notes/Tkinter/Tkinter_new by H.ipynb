```python
from tkinter import *
from tkinter import messagebox
from csv import writer
import os

# ----------------- FILE PATH -----------------
FILE_PATH = r"C:\Users\dearh\Downloads\imanshu 0321.csv"

# ----------------- CREATE WINDOW -----------------
window = Tk()
window.title("CSV Data Manager")
window.geometry("400x400")

# ----------------- CREATE CSV FILE IF NOT EXISTS -----------------
if not os.path.exists(FILE_PATH):
    with open(FILE_PATH, "w", newline="") as file:
        Writer = writer(file)
        Writer.writerow(["Name", "Age", "Mobile No", "Gender", "City"])

# ----------------- GLOBAL LIST -----------------
mainlist = []

# ----------------- LABELS & ENTRIES -----------------
Label(window, text="Name").grid(row=0, column=0, padx=10, pady=5)
Label(window, text="Age").grid(row=1, column=0, padx=10, pady=5)
Label(window, text="Mobile No").grid(row=2, column=0, padx=10, pady=5)
Label(window, text="Gender").grid(row=3, column=0, padx=10, pady=5)
Label(window, text="City").grid(row=4, column=0, padx=10, pady=5)

entry1 = Entry(window)
entry2 = Entry(window)
entry3 = Entry(window)
entry4 = Entry(window)
entry5 = Entry(window)

entry1.grid(row=0, column=1)
entry2.grid(row=1, column=1)
entry3.grid(row=2, column=1)
entry4.grid(row=3, column=1)
entry5.grid(row=4, column=1)

# ----------------- FUNCTIONS -----------------
def Add():
    if not validate_entries():
        return
    messagebox.showinfo("Info", "Data ready to save")

def Save():
    if not validate_entries():
        return

    temp_list = [
        entry1.get(),
        entry2.get(),
        entry3.get(),
        entry4.get(),
        entry5.get()
    ]

    if temp_list in mainlist:
        messagebox.showinfo("Info", "Data already exists")
        return

    mainlist.append(temp_list)

    with open(FILE_PATH, "a", newline="") as file:
        Writer = writer(file)
        Writer.writerow(temp_list)

    messagebox.showinfo("Info", "Data saved successfully")
    clear_entries()

def validate_entries():
    if (
        entry1.get() == "" or
        entry2.get() == "" or
        entry3.get() == "" or
        entry4.get() == "" or
        entry5.get() == ""
    ):
        messagebox.showwarning("Warning", "Please fill all the fields")
        return False
    return True

def clear_entries():
    entry1.delete(0, END)
    entry2.delete(0, END)
    entry3.delete(0, END)
    entry4.delete(0, END)
    entry5.delete(0, END)



# ----------------- BUTTONS -----------------
Button(window, text="Add", width=10, command=Add).grid(row=5, column=0, pady=10)
Button(window, text="Save", width=10, command=Save).grid(row=5, column=1)
Button(window, text="Quit", width=10, command=clear_entries).grid(row=6, column=0, columnspan=2)

# ----------------- MAINLOOP -----------------
window.mainloop()
```
