```python
#2
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount  #self.balance=self.balance + amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Invalid amount for deposit.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount #self.balance=self.balance - amount
            print(f"Withdrew ${amount}. New balance: ${self.balance}")
        else:
            print("Insufficient funds or invalid amount for withdrawal.")

    def get_balance(self):
        return self.balance

# Creating a BankAccount object
account = BankAccount("123456", 1000)

print("Initial balance:", account.get_balance())
account.deposit(500)
account.withdraw(200)
print("Final balance:", account.get_balance())
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass  # Abstract method, must be implemented in subclasses

    def breathe(self):
        print("This animal breathes air.")  # Concrete method

# Subclass that implements the abstract method
class Dog(Animal):

    def make_sound(self):
        print("Bark")

# Trying to instantiate Animal will raise an error
# animal = Animal()  # TypeError: Can't instantiate abstract class

# But you can instantiate Dog
dog = Dog()
dog.make_sound()  # Output: Bark
dog.breathe()     # Output: This animal breathes air.
```

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass

class WashingMachine(Appliance):
    def turn_on(self):
        print("Washing machine started.")

    def turn_off(self):
        print("Washing machine stopped.")

class Microwave(Appliance):
    def turn_on(self):
        print("Microwave heating...")

    def turn_off(self):
        print("Microwave turned off.")

obj=WashingMachine()
obj.turn_on()


obj=Microwave()
obj.turn_on()
```

```python
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started.")

    def stop_engine(self):
        print("Car engine stopped.")

class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine started.")

    def stop_engine(self):
        print("Bike engine stopped.")

c=Car()
c.start_engine()

b=Bike()
b.start_engine()
```

```python
class Vehicle(ABC):  #in this code data abstraction is achieved using encapsulation
    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

class Car(Vehicle):
    def __init__(self):
        self.__engine_temp = 75  # hidden internal data (private)==>__engine_temp is hidden from the user.

    def start_engine(self):
        print("Car engine started.")
        self.__engine_temp += 10  

    def stop_engine(self):
        print("Car engine stopped.")
        self.__engine_temp -= 5

    def get_engine_temp(self):
        return self.__engine_temp

c=Car()
c.start_engine()
#c.get_engine_temp()  #The user of Vehicle or Car doesn't need to know how engine_temp is managed internally.
c.stop_engine()
c.get_engine_temp()

#hidden data (for abstraction or encapsulation) is typically defined in the concrete class, not the abstract class.
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):  #The actual computation logic is hidden inside the Rectangle class.
    def __init__(self, length, width):
        self.length = length  # public data
        self.width = width    # public data

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

# User code only knows about Shape interface
shape: Shape = Rectangle(10, 5)
print("Area:", shape.area())         # 50
print("Perimeter:", shape.perimeter())  # 30
#The user doesn’t need to know how area() or perimeter() work.
#The user interacts only with the abstract interface (Shape), not the internal logic.
```

```python
 #Define the first parent class
class Parent1:
    def method1(self):
        print("Method from Parent1")

# Define the second parent class
class Parent2:
    def method2(self):
        print("Method from Parent2")

# Define the child class that inherits from both Parent1 and Parent2
class Child(Parent1, Parent2):
    def method3(self):
        print("Method from Child") 

# Create an instance of the Child class
child_instance = Child()

# Call methods from both parent classes and the child class
child_instance.method1()  # Inherited from Parent1
child_instance.method2()  # Inherited from Parent2
child_instance.method3()  # Defined in Child
```
