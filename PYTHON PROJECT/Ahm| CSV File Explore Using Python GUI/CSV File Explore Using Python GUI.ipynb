```python
from tkinter import *
from csv import *
from tkinter import messagebox
```

```python
window=Tk()
window.title("Simple| Data Entry")
```

```python
with open(r"C:\Users\dearh\Downloads\imanshu 0321.csv","w") as file:
        Writer=writer(file)  #writer function is from csv module
        Writer.writerow(["name","age","mo.no","gender","City"])
        
def Add():
    if not validate_entries():
        return
    messagebox.showinfo("Info", "Data ready to save")

def Save():
    temp_list = [entry1.get(), entry2.get(), entry3.get(), entry4.get(), entry5.get()]
    #=["kunal","pune"...],["ritesh"]
    global mainlist
    if 'mainlist' not in globals():
        mainlist = []
    # Check if the data already exists in mainlist
    if temp_list not in mainlist:
        mainlist.append(temp_list)
    # messagebox.showinfo("Info", "Data added successfully")
        with open(r"C:\Users\dearh\Downloads\imanshu 0321.csv","a",newline='') as file:
            Writer=writer(file)
            Writer.writerow(temp_list)
        messagebox.showinfo("Info", "Data saved successfully!")
    else:
        messagebox.showinfo("Info", "Data already exists") 

def validate_entries():
    if (
        entry1.get() == "" or
        entry2.get() == "" or
        entry3.get() == "" or
        entry4.get() == "" or
        entry5.get() == ""
    ):
        messagebox.showwarning("Warning", "Please fill all the fields")
        return False
    return True

def Exit():
    window.quit()
    window.destroy()
```

```python
def Clear():
    entry1.delete(0,END)
    entry2.delete(0,END)
    entry3.delete(0,END)
    entry4.delete(0,END)
    entry5.delete(0,END)
```

```python
label1=Label(window,text="Name: ", font=("Times New Roman", 20),bd=10, justify="right", padx=20,pady=20)
label2=Label(window,text="Age: ", font=("Times New Roman", 20),bd=10, justify="right", padx=20,pady=20)
label3=Label(window,text="Contact: ", font=("Times New Roman", 20),bd=10, justify="right", padx=20,pady=20)
label4=Label(window,text="Gender: ", font=("Times New Roman", 20),bd=10, justify="right", padx=20,pady=20)
label5=Label(window,text="City: ", font=("Times New Roman", 20),bd=10, justify="right", padx=20,pady=20)
entry1=Entry(window, font=("Times New Roman", 20),bd=10, width=30, borderwidth=5)
entry2=Entry(window, font=("Times New Roman", 20),bd=10, width=30, borderwidth=5)
entry3=Entry(window, font=("Times New Roman", 20),bd=10, width=30, borderwidth=5)
entry4=Entry(window, font=("Times New Roman", 20),bd=10, width=30, borderwidth=5)
entry5=Entry(window, font=("Times New Roman", 20),bd=10, width=30, borderwidth=5)
```

```python
add=Button(window, text="add", font=("Times New Roman", 20), bd=10, justify="right", padx=20, pady=20, command= Add)
save=Button(window, text="save", font=("Times New Roman", 20), bd=10, justify="right", padx=20, pady=20, command= Save)
clear=Button(window, text="clear", font=("Times New Roman", 20), bd=10, justify="right", padx=20, pady=20, command= Clear)
exit=Button(window, text="exit", font=("Times New Roman", 20), bd=10, justify="right", padx=20, pady=20, command= Exit)    
```

```python
label1.grid(row=0,column=0)
label2.grid(row=1,column=0)
label3.grid(row=2,column=0)
label4.grid(row=3,column=0)
label5.grid(row=4,column=0)
entry1.grid(row=0,column=1)
entry2.grid(row=1,column=1)
entry3.grid(row=2,column=1)
entry4.grid(row=3,column=1)
entry5.grid(row=4,column=1)
```
