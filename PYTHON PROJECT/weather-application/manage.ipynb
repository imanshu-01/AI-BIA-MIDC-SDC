## **Weather Application Using Python**

```python
# Import all Tkinter classes for GUI creation
from tkinter import *
# Import requests library to fetch data from API
import requests
# Import messagebox for showing pop-up messages (error, warning, info)
from tkinter import messagebox
```

```python
# ------------------ Function ------------------
def data_get(event=None):   # event for Enter key
    city = city_name.get().strip()
    if not city:
        messagebox.showwarning("Input Error", "Please enter a city name")
        return

    try:
        url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=68418ce436132483c0bce07db5b9f435&units=metric"
        data = requests.get(url, timeout=5).json()

        if data.get("cod") != 200:
            raise ValueError("City not found")

        desc_label1.config(text=data["weather"][0]["description"].title())
        temp_label1.config(text=f"{data['main']['temp']} °C")
        press_label1.config(text=f"{data['main']['pressure']} hPa")
        hum_label1.config(text=f"{data['main']['humidity']} %")
        wind_label1.config(text=f"{data['wind']['speed']} m/s")

    except:
        messagebox.showerror("Error", "City not found or Internet issue")
```

```python
# ------------------ Window ------------------
win = Tk()
win.title("Advanced Weather App")
win.geometry("500x580")
win.config(bg="#87CEEB")
```

```python
# ------------------ Heading ------------------
Label(
    win, text="Weather Application",
    font=("Times New Roman", 28, "bold"),
    bg="#87CEEB"
).place(x=25, y=30, width=450)
```

```python
# ------------------ City Entry (Keyboard Input) ------------------
city_name = StringVar()

city_entry = Entry(
    win, textvariable=city_name,
    font=("Times New Roman", 18),
    justify="center"
)
city_entry.place(x=25, y=100, width=450, height=45)
city_entry.focus()  # cursor active

# Press ENTER to get weather
city_entry.bind("<Return>", data_get)
```

```python
# ------------------ Button ------------------
Button(
    win, text="Get Weather",
    font=("Times New Roman", 18, "bold"),
    bg="#1E90FF", fg="white",
    command=data_get
).place(x=170, y=160, width=160, height=45)
```

```python
# ------------------ Row Function ------------------
def create_row(text, y):
    Label(
        win, text=text,
        font=("Times New Roman", 18),
        bg="white"
    ).place(x=25, y=y, width=210, height=45)

    lbl = Label(
        win, text="",
        font=("Times New Roman", 18),
        bg="#F0F8FF"
    )
    lbl.place(x=250, y=y, width=210, height=45)
    return lbl
```

```python
# ------------------ Weather Details ------------------
desc_label1 = create_row("Weather Description", 230)
temp_label1 = create_row("Temperature", 290)
press_label1 = create_row("Pressure", 350)
hum_label1 = create_row("Humidity", 410)
wind_label1 = create_row("Wind Speed", 470)

win.mainloop()
```
