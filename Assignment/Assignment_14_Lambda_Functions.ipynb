### 1. What is a lambda function in Python?

A lambda function is a small anonymous function written in one line. It is used for short operations.

### 2. What is the syntax for defining a lambda function?

```python
square = lambda x: x * x
print(square(5))
```

### 3. Can a lambda function have multiple parameters?

```python
add = lambda a, b: a + b
print(add(3, 4))
```

### 4. What is the difference between a lambda function and regular function?

Lambda is one-line and has no name. Regular functions use def and can have multiple lines.

### 5. Can a lambda function contain multiple expressions?

No, lambda functions can have only one expression.

### 6. When is it appropriate to use a lambda function instead of a regular function?

Use lambda when the function is small and used only once.

### 7. How do you use the map() function with a lambda function?

```python
numbers = [1, 2, 3, 4]
result = list(map(lambda x: x * 2, numbers))
print(result)
```

### 8. What is the purpose of the filter() function when used with a lambda function?

```python
numbers = [1, 2, 3, 4, 5]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)
```

### 9. Can you use a lambda function to calculate the sum of elements in a list?

```python
from functools import reduce
numbers = [1, 2, 3, 4]
total = reduce(lambda a, b: a + b, numbers)
print(total)
```

### 10. How do you create a lambda function to find the square of a number?

```python
square = lambda x: x * x
print(square(6))
```

### 11. How do you use lambda functions with the sorted() function to sort dictionaries?

```python
students = {'A': 85, 'B': 70, 'C': 90}
sorted_students = sorted(students.items(), key=lambda x: x[1])
print(sorted_students)
```

### 16. Develop a program that calculates the sum of elements in a list using a lambda function with the reduce() function from the functools module.
[Note: The reduce() function is used when you want to perform a

sequence to a single value. It's particularly useful when you have a series of
values and you want to combine them in a specific way to obtain a final

### 26. Write a program that checks whether a list of numbers are prime or not using a lambda function.

```python
numbers = [2, 3, 4, 5]
is_prime = lambda n: n > 1 and all(n % i != 0 for i in range(2, n))
print(list(map(is_prime, numbers)))
```
