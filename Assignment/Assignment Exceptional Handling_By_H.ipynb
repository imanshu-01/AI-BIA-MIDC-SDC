## 11. Create a banking application simulation that throws exceptions for insufficient balance, invalid account numbers, and wrong PIN. Define custom exceptions.

```python
def withdraw(acc, pin, amt):
    balance = 5000
    correct_acc = 101
    correct_pin = 1234

    if acc != correct_acc:
        raise Exception("Invalid Account Number")

    if pin != correct_pin:
        raise Exception("Wrong PIN")

    if amt > balance:
        raise Exception("Insufficient Balance")

    return balance - amt

try:
    acc = int(input("Enter Account Number: "))
    pin = int(input("Enter PIN: "))
    amt = int(input("Enter Amount: "))
    
    print("Remaining Balance:", withdraw(acc, pin, amt))

except Exception as e:
    print("Error:", e)
```

## 12. Accept marks for 5 subjects from the user. Raise an exception if: The input is not a number and any mark is outside the 0–100 range. Then compute the average and grade. Use exception handling to manage errors gracefully.

```python
try:
    total = 0

    for i in range(1, 6):
        m = int(input("Enter marks of subject:"))

        if m < 0 or m > 100:
            raise ValueError
    total += m
    avg = total / 5

    if avg >= 40:
        print("Average =", avg)
        print("Result: Pass")
    else:
        print("Average =", avg)
        print("Result: Fail")

except ValueError:
    print("Error: Enter marks between 0 and 100 only")
```

## 13. Create a shopping cart program where the user enters product prices and quantities. Ensure the program handles: Invalid number or price inputs. Negative prices or quantities. Calculate and print the total cart value. Use try–except–finally blocks.

```python
total = 0
try:
    while True:
        price = float(input("Enter price: "))
        if price == 0:
            break
        qty_value = int(input("Enter quantity: "))
        if price < 0 or qty_value < 0:
            raise ValueError
        total += price * qty_value
except ValueError:
    print("Invalid input")
finally:
    print("Total cart value:", total)
```

## 14. Define a custom exception called InvalidEmailError. Write a function that checks if an input string is a valid email (must contain @ and . and no spaces). If invalid, raise InvalidEmailError. Use try–except to test it with various inputs.

```python
class InvalidEmailError(Exception): pass

def check_email(email):
    if "@" not in email or "." not in email or " " in email:
        raise InvalidEmailError("Invalid Email")
    return "Valid Email"

try:
    print(check_email("dearhimanshu3@gmail.com"))
except InvalidEmailError as e:
    print(e)
```

## 15. Build a number guessing game. The player gets 3 chances to guess a randomly generated number between 1 and 10. If the input is not a number, catch the exception and don’t count it as a turn. Show the number of valid attempts made and whether the player won or lost.

```python
import random
number = random.randint(1,10)
attempts = 0

while attempts < 3:
    try:
        guess = int(input("Guess number: "))
        attempts += 1
        if guess == number:
            print("Booyah!")
            break
    except ValueError:
        print("Invalid input")
else:
    print("You lost! Number was:", number)
```

## 17. Simulate a login system with hardcoded username and password. The user has 3 attempts. Raise InvalidLoginError (custom exception) on incorrect login. Lock the account if attempts exceed 3. Use exception handling to manage logic cleanly.

```python
class InvalidLoginError(Exception): pass

username = "imanshu"
password = "1253"

for i in range(3):
    try:
        u = input("Username: ")
        p = input("Password: ")
        if u != username or p != password:
            raise InvalidLoginError("Invalid login")
        print("Login successful")
        break
    except InvalidLoginError as e:
        print(e)
else:
    print("Account locked")
```

## 18. Create a program that takes a file name and a new name as input. It tries to rename the file. Handle cases where: The original file does not exist. The new file name already exists. Permission is denied.

```python
import os

original_file = input("Enter the original file name: ")
new_file = input("Enter the new file name: ")

try:
    os.rename(original_file, new_file)
    print("File renamed successfully from '" + original_file + "' to '" + new_file + "'.")

except FileNotFoundError:
    print("Error: The file '" + original_file + "' does not exist.")

except FileExistsError:
    print("Error: The file '" + new_file + "' already exists.")

except PermissionError:
    print("Error: Permission denied. You cannot rename this file.")

except Exception as e:
    print("An unexpected error occurred: " + str(e))
```
